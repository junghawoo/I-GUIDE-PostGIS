```python
import geopandas as gpd
import pandas as pd
from pathlib import Path
```

```python
import os
```

```python
# --- Configuration ---

root_dir = Path("/Users/xiaoliu/Work/Project/I-GUIDE/IGUIDE_Aging_Dam-selected/AgingDamsPopulationData/US_Census_Tracts/")
# Adjust the pattern to search for shapefiles in subdirectories
shapefile_pattern = "**/*.shp" 

# --- Process ---
all_files = list(root_dir.glob(shapefile_pattern))
```

```python
gdfs = []

for shp_path in all_files:
    # Get the Path object for the subdirectory
    subdirectory_path = shp_path.parent    
    # Get the name (string) of the subdirectory, e.g., 'A_folder'
    subdirectory_name = subdirectory_path.name

    state_name = subdirectory_name.split('_')[2]
    print(f"State: {state_name}")

    

    try:
       # You can add a new column here to store the source filename/directory if needed
       gdf = gpd.read_file(shp_path)

       gdf['state'] = state_name
            
       gdfs.append(gdf)
    except Exception as e:
       print(f"Could not read {shp_path}: {e}")

  # Concatenate all GeoDataFrames
    # The `ignore_index=True` creates a new continuous index
    # Note: If CRS differs, you'll need to handle that, e.g., by reprojecting before concatenation
    merged_gdf = pd.concat(gdfs, ignore_index=True)

    # Convert the resulting pandas DataFrame back to GeoDataFrame if needed (though concat on GDFs usually preserves it)
    if not isinstance(merged_gdf, gpd.GeoDataFrame):
        merged_gdf = gpd.GeoDataFrame(merged_gdf, geometry='geometry', crs=gdfs[0].crs)
        #merged_gdf = merged_gdf.to_crs(epsg=4326)
        
    print(f"Successfully merged {len(gdfs)} shapefiles into a GeoDataFrame with {len(merged_gdf)} features.")
```

```python
# Let's assume both dataframes have a common column named 'ID'
common_key = 'GEOID' 

joined_gdf = merged_gdf.merge(
    acsd_keep, 
    on=common_key, 
    how='left'  # or 'inner', 'right', 'outer'
)

joined_gdf2 = joined_gdf.merge(
    ruca_merge,
    left_on='GEOID',
    right_on='TractFIPS23',
    how='left'
)

final_gdf = joined_gdf2.merge(
    svi,
    left_on='GEOID',
    right_on='FIPS',
    how='left'
)

print(f"Original columns: {list(merged_gdf.columns)}")
print(f"Final columns (joined): {list(final_gdf.columns)}")
```

```python
final_gdf.head()
```

```python
target_crs = "EPSG:4326"
reprojected_gdf = final_gdf.to_crs(target_crs)

print(f"Original CRS: {merged_gdf.crs}")
print(f"New CRS: {reprojected_gdf.crs}")
```
