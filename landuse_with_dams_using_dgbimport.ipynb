This jupyter notebook needs "landuse2024"(generated from preprocess/landuse.txt) and "utah_dam_inundation_zones" layers in the postgis database.

The three cells below test different projection (4326, 29912, and North American Albers Equal Area 102008). The last cell (North American Albers Equal Area 102008) is the most accurate one and should be used.

```python
import folium
from folium import plugins
import psycopg2
import json
import branca.colormap as cm


# ----------------------------
# Connect to PostGIS
# ----------------------------
conn = psycopg2.connect(
    dbname="utahdaminundationprofiles_aug9_2025",
    user="admin",
    password="admin",
    host="localhost",
    port=5432
)
cur = conn.cursor()

# ----------------------------
# Create folium map
# ----------------------------
#m = folium.Map(location=[40.7607, -111.8939], zoom_start=7)

# ----------------------------
# Query all dams (including 0-intersection)
# ----------------------------
cur.execute("""

SELECT 
    objectid,
    damnumber,
    name,
    (stats).value,
    SUM((stats).count) AS total_cells
FROM (
    SELECT 
        d.objectid,
        d.damnumber,
        d.name,
        ST_ValueCount(ST_Clip(r.rast, d.geom)) AS stats
    FROM landuse2024 r
    INNER JOIN utah_dam_inundation_zones d 
        ON ST_Intersects(r.rast, d.geom)
) AS subquery
GROUP BY objectid, damnumber, name, (stats).value
ORDER BY objectid, (stats).value;

""")
#LIMIT 50
results = cur.fetchall()





# ----------------------------
# Cleanup
# ----------------------------
cur.close()
conn.close()

```

```python
import folium
from folium import plugins
import psycopg2
import json
import branca.colormap as cm


# ----------------------------
# Connect to PostGIS
# ----------------------------
conn = psycopg2.connect(
    dbname="utahdaminundationprofiles_aug9_2025",
    user="admin",
    password="admin",
    host="localhost",
    port=5432
)
cur = conn.cursor()

# ----------------------------
# Create folium map
# ----------------------------
#m = folium.Map(location=[40.7607, -111.8939], zoom_start=7)

# ----------------------------
# Query all dams (including 0-intersection)
# ----------------------------
cur.execute("""

SELECT 
    objectid,
    damnumber,
    name,
    (stats).value,
    SUM((stats).count) AS total_cells
FROM (
    SELECT 
        d.objectid,
        d.damnumber,
        d.name,
        ST_ValueCount(ST_Transform(ST_Clip(r.rast, d.geom), 102008)) AS stats
    FROM landuse2024 r
    INNER JOIN utah_dam_inundation_zones d 
        ON ST_Intersects(r.rast, d.geom)
) AS subquery
GROUP BY objectid, damnumber, name, (stats).value
ORDER BY objectid, (stats).value;

""")
#LIMIT 50
results3 = cur.fetchall()





# ----------------------------
# Cleanup
# ----------------------------
cur.close()
conn.close()

```
